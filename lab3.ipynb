Problem 3: Checking Routes Between Cities Using DFS Problem Statement: A small road network connects several cities. Write a Python program using DFS that:

Checks whether a route exists between two given cities.
Finds all separate groups of cities that are connected to each other but not to any other group (these are called connected components).

```python
# DFS to check route between cities
# and find connected components

def dfs(city, graph, visited, component):
    visited.add(city)
    component.append(city)

    for neighbour in graph[city]:
        if neighbour not in visited:
            dfs(neighbour, graph, visited, component)


def route_exists(graph, start, destination):
    visited = set()

    def search(city):
        if city == destination:
            return True

        visited.add(city)

        for neighbour in graph[city]:
            if neighbour not in visited:
                if search(neighbour):
                    return True

        return False

    return search(start)


# Road network
graph = {
    "Kolkata": ["Delhi", "Mumbai"],
    "Delhi": ["Kolkata", "Mumbai"],
    "Mumbai": ["Kolkata", "Delhi"],
    "Chennai": ["Pune"],
    "Pune": ["Chennai"],
    "Jaipur": []
}


# Part 1: Check whether a route exists
start = "Kolkata"
destination = "Mumbai"

if route_exists(graph, start, destination):
    print("Route exists from", start, "to", destination)
else:
    print("No route exists from", start, "to", destination)


# Part 2: Find connected components
visited = set()
components = []

for city in graph:
    if city not in visited:
        component = []
        dfs(city, graph, visited, component)
        components.append(component)


print("\nConnected Components:")

for i, component in enumerate(components, 1):
    print("Component", i, ":", component)
```
