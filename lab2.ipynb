Problem 2: The Water Jug Problem (Solved using BFS) Problem Statement: You are given two jugs: Jug A with a capacity of 4 litres, and Jug B with a capacity of 3 litres. Neither jug has any markings on it. Using only the following operations — (i) fill a jug completely, (ii) empty a jug completely, (iii) pour water from one jug into the other until either the first jug is empty or the second jug is full — determine the minimum sequence of steps needed to measure out exactly 2 litres of water in any one jug.

```python
from collections import deque

A = 4
B = 3

start = (0, 0)

queue = deque([(start, [])])
visited = {start}

while queue:
    (a, b), path = queue.popleft()

    if a == 2 or b == 2:
        print("Initial State: (0, 0)")

        for i, (operation, state) in enumerate(path, 1):
            print(f"Step {i}: {operation}")
            print(f"State: {state}")

        print("\nTotal steps:", len(path))
        break

    # Fill Jug A
    state = (A, b)
    if state not in visited:
        visited.add(state)
        queue.append((state, path + [("Fill Jug A", state)]))

    # Fill Jug B
    state = (a, B)
    if state not in visited:
        visited.add(state)
        queue.append((state, path + [("Fill Jug B", state)]))

    # Empty Jug A
    state = (0, b)
    if state not in visited:
        visited.add(state)
        queue.append((state, path + [("Empty Jug A", state)]))

    # Empty Jug B
    state = (a, 0)
    if state not in visited:
        visited.add(state)
        queue.append((state, path + [("Empty Jug B", state)]))

    # Pour A into B
    amount = min(a, B - b)
    state = (a - amount, b + amount)
    if state not in visited:
        visited.add(state)
        queue.append((state, path + [("Pour Jug A into Jug B", state)]))

    # Pour B into A
    amount = min(b, A - a)
    state = (a + amount, b - amount)
    if state not in visited:
        visited.add(state)
        queue.append((state, path + [("Pour Jug B into Jug A", state)]))
```
